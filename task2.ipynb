```python
class calc:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        return self.a + self.b

    def subtract(self):
        return self.a - self.b

    def multiply(self):
        return self.a * self.b

    def divide(self):
        if self.b != 0:
            return self.a / self.b
        else:
            return "Error! Division by zero."
```

```python
def calculator():
    print("Welcome to the calculator!")
    print("Choose an operation:")
    print("1: Add")
    print("2: Subtract")
    print("3: Multiply")
    print("4: Divide")

    choice = input("Enter the number of the operation you want to perform (1/2/3/4): ")

    if choice in ['1', '2', '3', '4']:
        num1 = float(input("Enter the first number: "))
        num2 = float(input("Enter the second number: "))

        my_calc = calc(num1, num2)

        if choice == '1':
            print(f"{num1} + {num2} = {my_calc.add()}")
        elif choice == '2':
            print(f"{num1} - {num2} = {my_calc.subtract()}")
        elif choice == '3':
            print(f"{num1} * {num2} = {my_calc.multiply()}")
        elif choice == '4':
            result = my_calc.divide()
            print(f"{num1} / {num2} = {result}")
    else:
        print("Invalid choice!")



```

```python
calculator()
```
