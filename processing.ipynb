```python
def Drop_unnecessary_features(df, cols_to_drop):
    df = df.drop(columns=cols_to_drop, errors='ignore')
    return df


cols_to_drop = ['PassengerId', 'Name', 'Ticket']


df = Drop_unnecessary_features(df, cols_to_drop)


df.head()
```

```python
def Check_data_type(df):
    report = pd.DataFrame({
        'Datatype': df.dtypes,
        'Unique_Values': df.nunique()
    })
   
    return report.T
```

```python
Check_data_type(df)
```

```python

def check_outliers_iqr(df):
    num_col = df.select_dtypes("number").columns
    
    for col in num_col:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_fence = Q1 - 1.5 * IQR
        upper_fence = Q3 + 1.5 * IQR
        
        print(f"Column: {col}")
        print(f"Q1: {Q1}, Q3: {Q3}")
        print(f"IQR: {IQR}")
        print(f"Bounds: ({lower_fence}, {upper_fence})")
        print("-" * 30)
```

```python
check_outliers_iqr(df)
```
