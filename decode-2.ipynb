```python
message = "&&&**$gnirtS PLIO!!@1234"

core = ""
for ch in message:
    if ch.isalpha() or ch.isspace():
        core += ch

core = core.strip() 


word1, word2 = core.split()


decoded_word1 = word1[::-1]

decoded_word2 = word2.replace("I", "E").replace("O", "U")

```

```python
print(decoded_word1, decoded_word2)
```
