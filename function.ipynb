```python
def multiplication_table(num):
           
    for i in range(1, 11):
            result = num * i
            print(f"{num} * {i} = {result}")
```

```python
multiplication_table(9)
```

```python
def primeFactors(n):
    factors = []
    divisor = 2

    while n > 1:
        if n % divisor == 0:
            factors.append(divisor)
            n = n // divisor
        else:
            divisor += 1

    return factors
```

```python

print(primeFactors(56))
```

```python
def decToBin(n):
    binary = ""

    while n > 0:
        remainder = n % 2
        binary = str(remainder) + binary
        n = n // 2

    return binary
```

```python
print(decToBin(11))
```
