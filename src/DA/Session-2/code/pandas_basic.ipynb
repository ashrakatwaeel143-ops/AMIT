```python
import numpy as np
def convert_dtype(arr , dtype):
    try:
        arr = np.asarray(arr)
        return arr.astype(dtype)
    except TypeError as r:
        raise TypeError(f"Invalid target data type {dtype}") from r
    except ValueError as r:
            raise ValueError(f"can not convert to {dtype}") from r
    
lst = [1.5 , 2.9 , 3.1 ]
convert_dtype(lst , int )
```

```python
def check_dtype(arr , dtype):
    if not isinstance(arr , np.ndarray):
        print("input must be a numpy")
    dtype = np.dtype(dtype)
    return arr.dtype == dtype
```

```python
lst = [1 , 2 , 3 ]
check_dtype(lst , int )
```

```python
def init_weights(n_feature, method = "zeros", seed =None):
    if not isinstance(n_feature, int):
        raise TypeError(f"{n_feature} must be an integer")
    if n_feature <= 0 :
        raise TypeError(f"{n_feature} must be gretar than zero")
    if method == "zeros":
        return np.zeros(n_feature)
    if method == "random":
        if seed is not None:
            return np.random.randn(n_feature) 
    raise  ValueError(f"un support the method type{method}")
print(init_weights(4,"random",42))
print(init_weights(4,"zeros"))
# print(init_weights(4.1,"zeros"))
```
