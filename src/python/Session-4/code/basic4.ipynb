```python
def swap(word: str) -> list:
    '''
    This function takes a string as input and returns a list of words in reverse order.
    Args:
        word (str): The input string to swap.
    Returns:
        list: A list of words in reverse order.
    Example:
        >>> swap("Osama Mohammed")
        ['Mohammed', 'Osama']
    '''
    lst = word.split(" ")
    lst.reverse()
    return lst
word=input("Enter a word to swap: ")
result = swap(word)
print(result
```
