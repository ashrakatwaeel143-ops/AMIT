```python
def is_prime(n):
       if n < 2:
           return False
       for i in range(2, n):
           if n % i == 0:
               return True
for j in range(3, 1000, 2):
    if is_prime(j) and is_prime(j + 2):
        print(j, "and", j + 2)
     
```

```python
def primeFactors(n):
    factors = []
    divisor = 2

    while n > 1:
        if n % divisor == 0:
            factors.append(divisor)
            n = n // divisor
        else:
            divisor += 1

    return factors
```

```python
print(primeFactors(56))
```

```python
def decToBin(n):
    binary ="" 

    while n > 0:
        remainder = n % 2
        binary = str(remainder) + binary
        n = n // 2

    return binary


print(decToBin(11))
```

```python
def perfectNums(start, end):
    for num in range(start, end + 1):
        if num == 0:
            print(num)  
           

        divisors_sum = 0
        for i in range(1, num):
            if num % i == 0:
                divisors_sum += i

        if divisors_sum == num:
            print(num)
```

```python
perfectNums(0, 100)
```
