```python
class TupleModifier:
    """
    TupleModifier class handles adding an item to a tuple
    
    Attr: my_tuple (tuple): The original tuple
        item_to_add: The item to add to the tuple
    """
    def __init__(self, my_tuple, item_to_add):
        self.my_tuple = my_tuple
        self.item_to_add = item_to_add

    def add_item(self):
        temp_list = list(self.my_tuple)
        temp_list.append(self.item_to_add)
        return tuple(temp_list)

obj = TupleModifier((1, 4, 5), 4)
print(obj.add_item())
```

```python
class ListSummer:

    def __init__(self, numbers):
        self.numbers = numbers

    def calculate_total(self):
        """
        Loops through the list and calculates the total sum
        
           Ret : int/float: The sum of all item in the list
        """
        total = 0
        for num in self.numbers:
            total += num
        return total


obj = ListSummer([5, 10, 15, 20])
print(obj.calculate_total())
```

```python
class ListMultiplier:
    """
    ListMultiplier class handles multiplying all items in a list.
    Attributes:
        numbers (list): The list of numbers to multiply.
    """
    def __init__(self, numbers):
        self.numbers = numbers

    def calculate_product(self):
        """
        Loops through the list and multiplies all items together.
        Returns:
            int/float: The product of all items in the list.
        """
        result = 1
        for num in self.numbers:
            result *= num
        return result


obj = ListMultiplier([100, 100, 100])
print(obj.calculate_product())
```

```python
class SmallestFinder:
    """
    Smallestfinder class handles finding the smallest number in a list manually
  
    """
    def __init__(self, numbers):
        self.numbers = numbers

    def find_smallest(self):    
        """
        Loops through the list and finds the smallest value without using min().
        Returns:
            int/float: The smallest number in the list.
        """
        smallest = self.numbers[0]
        for num in self.numbers:
            if num < smallest:
                smallest = num
        return smallest

obj = SmallestFinder([10, 3, 25, 1, 7,0.1])
print(obj.find_smallest())
```

```python
class LargestFinder:

    def __init__(self, numbers):
        self.numbers = numbers

    def find_largest(self):
            """
            Loops through the list and finds the largest value without using max().
            Returns:
                int/float: The largest number in the list.
            """
            largest = self.numbers[0]
            for num in self.numbers:
                if num > largest:
                    largest = num
            return largest
    

obj = LargestFinder([10, 3, 25, 1, 7])
print(obj.find_largest())
```

```python
class StringLengthCounter:
    """
    StringLengthCounter class handles counting strings in a list with length >= 2
   
       Atr: words The list of strings to check.
    """
    def __init__(self, words):
        self.words = words

    def count_long_words(self):
            """
            Loops through the list and counts strings with length >= 2.
            Returns:
                int: The count of strings with length >= 2.
            """
            count = 0
            for word in self.words:
                if len(word) >= 2:
                    count += 1
            return count

obj = StringLengthCounter(["hi", "a", "hello", "no", "on","Ashrakat"])
print(obj.count_long_words())
```

```python
class ListCloner:
    def __init__(self, original):
        self.original = original

    def clone_list(self):
         """
         Creates a real copy of the list (not a reference).
         Returns:
             list: A new independent copy of the original list.
         """
         cloned = self.original[:]
         return cloned

obj = ListCloner([1, 2, 3, 4])
print(obj.clone_list())
```

```python
class SetItemRemover:
    def __init__(self, my_set):
        self.my_set = my_set

    def remove_item(self, item):
          """
          Removes the given item from the set using discard() (safe even if not present).
          Returns:
              set: The set after removing the item.
          """
          self.my_set.discard(item)
          return self.my_set

obj = SetItemRemover({1, 2, 3, 4, 5})
print(obj.remove_item(3))
print(obj.remove_item(1))
```

```python
class SubsetChecker:
    """
    SubsetChecker class handles checking whether one set is a subset of another
    
    """
    def __init__(self, set1, set2):
        self.set1 = set1
        self.set2 = set2

    def check_subset(self):
            """
            Checks whether set1 is a subset of set2 using issubset().
            Returns:
                bool: True if set1 is a subset of set2, otherwise False.
            """
            return self.set1.issubset(self.set2)

obj = SubsetChecker({1, 2}, {1, 2, 3, 4})
print(obj.check_subset())
```
