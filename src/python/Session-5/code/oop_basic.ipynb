```python
class clac:
    def __init__(self,x,y)->None:
        self.x=x
        self.y=y
    def add(self):
        return self.x + self.y

c1=clac(4,6)
print(c1.add())
```

```python
from abc import ABC,abstractmethod
class Animal(ABC):
    @abstractmethod
    def move(self):
        pass

class Bird(Animal):
    def move(self):
        print("Bird in Animal")
        pass

class Cat(Bird):
    def move(self):
        pass
```

```python
b1 = Cat()
print(b1.move())
```

```python
class parent:
    def __init__(self)-> None:
        self.__value = 10

class child(parent):
    def __init__(self)->None:
        super().__init__()
        self.__value = 20

c=child()
print(c.__dict__)
```

```python
from abc import ABC,abstractmethod

class A():
    def do_this(self):
        print("A")


class B(A):
    pass


class C(ABC):
    @abstractmethod
    def do_this(self):
        print("C")


class D(B,C):
    def do_this(self):
        return super().do_this()


x=D()
x.do_this()
```
