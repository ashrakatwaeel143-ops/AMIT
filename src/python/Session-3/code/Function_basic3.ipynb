```python
def train_model(model, *data, **hyperparameters):
    print(f"Training model: {model}")
    print("\nDatasets:")
    for dataset in data:
        print(f"- {dataset}")
    print("\nHyperparameters:")
    for key, value in hyperparameters.items():
        print(f"  {key}: {value}")

train_model(
    "fraids_detection_model",
    "train.csv",
    "validation.csv",
    "test.csv",
    learning_rate=0.01,
    batch_size=32,
    
    epochs=10,
    optimizer="adam",
    earlystopping=True

    )
```
