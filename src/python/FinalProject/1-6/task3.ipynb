```python
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name):
        self.name = name

   
    def make_sound(self):
        pass

    def describe(self):
        print(f"{self.name} is an animal.")

```

```python
class Dog(Animal):
    def make_sound(self):
        return "Woof!"
```

```python
class Cat(Animal):
    def make_sound(self):
        return "Meow!"
```

```python
class Cow(Animal):
    def make_sound(self):
        return "Moo!"
```

```python
# Usage
animals = [Dog("Rex"), Cat("Whiskers"), Cow("Bessie")]

for animal in animals:
    animal.describe()
    print(f"{animal.name} says: {animal.make_sound()}")
    print()
```
