```python
name = "Debi of Machine Learning"

print("Upper case:", name.upper())
print("Lower case:", name.lower())
print("Title case:", name.title())
print("Capitalize:", name.capitalize())
print("Replace:", name.replace("Machine Learning", "Deep Learning"))
print("Split:", name.split())
print("Split 'of':", name.split("of"))
print("Find 'Machine':", name.find("Machine"))
print("Starts with 'of':", name.startswith("of"))
print("Count 'e':", name.count("e"))

```
