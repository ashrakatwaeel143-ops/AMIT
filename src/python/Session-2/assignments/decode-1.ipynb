```python
message = "###!!@mocleW EPGTQ!!!6789"


core = ""
for ch in message:
    if ch.isalpha() or ch.isspace():
        core += ch

core = core.strip() 


word1, word2 = core.split()


decoded_word1 = word1[::-1]


decoded_word2 = word2
```

```python
print(decoded_word1, decoded_word2)
```

```python
vowels = "AE"

if word2[0] in vowels:
    decoded_word2 = word2[1:]
else:
    decoded_word2 = word2
 
```

```python
print(decoded_word1, decoded_word2)
```
