```python
def check_email(email):
 
    if email.count("@") != 1:
        return "Invalid email"
 
    at_index = email.index("@")
 
   
    if "." not in email[at_index:]:
        return "Invalid email"
 
    username = email.split("@")[0]
    domain = email.split("@")[1].rsplit(".", 1)[0]
 
    if email.endswith(".com"):
        ending = "Commercial Domain"
    elif email.endswith(".edu"):
        ending = "Educational Domain"
    else:
        ending = "Other Domain"
 
    print("Username:", username)
    print("Domain:", domain)
    print("Type:", ending)
 
 
```

```python

check_email("Amit_ml@gmail.edu")
```
